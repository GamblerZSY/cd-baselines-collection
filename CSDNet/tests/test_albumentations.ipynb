```python
import albumentations as A
import cv2
from matplotlib import pyplot as plt
import os
```

```python
def visualize(image):
    plt.figure(figsize=(10, 10))
    plt.axis("off")
    plt.imshow(image)
```

```python
path = '/media/jic2/HDD/DSJJ/CDdata/LEVIR-CD/train'
name = 'train_9.png'

imgA = cv2.imread(os.path.join(path, 'A', name), cv2.IMREAD_COLOR_RGB) 
imgB = cv2.imread(os.path.join(path, 'B', name), cv2.IMREAD_COLOR_RGB)
lab = cv2.imread(os.path.join(path, 'label', name), cv2.IMREAD_COLOR_RGB)
```

```python
visualize(imgA)
visualize(imgB)
```

```python
crop_size = 256
transform = A.Compose(
    A.RandomCrop(height=crop_size, width=crop_size),
    additional_targets={"image1": "image", "mask": "mask"},
    strict=True,
    seed=120,
)
# transform = A.Compose(
#     [A.VerticalFlip(p=1)],
#     additional_targets={"image0": "image", "image1": "image"},
#     strict=True,
#     seed=137,
# )
transformed = transform(image=imgA, image1=imgB, mask=lab)
```

```python
visualize(transformed["image"])
visualize(transformed["image1"])
```
